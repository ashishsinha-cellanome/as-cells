# Mask2Former

```python
from mask2former_model import Mask2FormerInstanceSegmentation
from AbstractVisionModel import run_model

import cv2
import os
import numpy as np
from PIL import Image
from typing import List, Dict, Union, Tuple, Final
```

```python
MODEL_WEIGHTS_PATH: Final[str] = '/home/cellareye/Cellanome/dl-mehdi/Mask RCNN/checkpoints/mask2former_checkpoints/20250312_mask2former_sets_1_2_3_6_to_41_0p1_bbox_0p7_1_rs_0p25_blur_2_bs_8_epochs_1cl_lrs.pt'
detector = Mask2FormerInstanceSegmentation(weights_path=MODEL_WEIGHTS_PATH, adjust_masks=True)
print(detector.get_metadata())
```

```python
img = cv2.imread('/home/cellareye/Cellanome/dl-mehdi/COCO Evaluation/7_4_6_4851_1_008526_035728_-16726_White.png', cv2.IMREAD_UNCHANGED)
```

```python
detections, runtime, debubg_img = run_model(
    detector=detector,
    input_image=img,
    normalize_image=True,
    bit_depth=12,
    post_process_class_names=list(detector.get_reverse_label_map().keys()),
    plot_results=True
)
```

```python
classname_to_class_id_map = detector.get_reverse_label_map()
for classname, class_id in classname_to_class_id_map.items():
    print(f"Number of {classname}s: {len([label for label in detections['labels'] if label == class_id])}")
Image.fromarray(debubg_img)
```
